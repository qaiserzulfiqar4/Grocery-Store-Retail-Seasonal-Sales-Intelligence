```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Grocery Seasonal Sales Analysis")
print("Libraries imported successfully!")
```

```python
np.random.seed(42)

n_orders = 1200

dates = pd.date_range(
    start="2025-01-01",
    end="2025-12-31",
    periods=n_orders
)

products = {
    "Rice": {"price": 280, "cost": 220},
    "Milk": {"price": 220, "cost": 175},
    "Sugar": {"price": 160, "cost": 130},
    "Cooking Oil": {"price": 600, "cost": 510},
    "Flour": {"price": 150, "cost": 120},
    "Tea": {"price": 850, "cost": 720},
    "Biscuits": {"price": 100, "cost": 70},
    "Eggs": {"price": 300, "cost": 240}
}

product_names = list(products.keys())

month_factors = {
    1: 0.90, 2: 0.95, 3: 1.00, 4: 1.05,
    5: 1.15, 6: 1.20, 7: 1.10, 8: 1.05,
    9: 0.90, 10: 0.95, 11: 1.00, 12: 1.30
}

rows = []

for i, date in enumerate(dates):
    product = np.random.choice(product_names)
    quantity = np.random.randint(1, 11)

    base_price = products[product]["price"]
    base_cost = products[product]["cost"]

    unit_price = round(
        base_price * np.random.uniform(0.95, 1.05), 2
    )
    unit_cost = round(
        base_cost * np.random.uniform(0.97, 1.03), 2
    )

    seasonal_factor = month_factors[date.month]
    quantity = max(1, round(quantity * seasonal_factor))

    revenue = round(quantity * unit_price, 2)
    cost = round(quantity * unit_cost, 2)
    profit = round(revenue - cost, 2)

    rows.append([
        f"ORD{i+1:04d}", date, product, quantity,
        unit_price, unit_cost, revenue, cost, profit
    ])

df = pd.DataFrame(rows, columns=[
    "Order_ID", "Date", "Product", "Quantity",
    "Unit_Price", "Unit_Cost", "Revenue", "Cost", "Profit"
])

print("Sample dataset created successfully!")
```

```python
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.strftime("%b")

def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4]:
        return "Spring"
    elif month in [5, 6]:
        return "Summer"
    elif month in [7, 8, 9]:
        return "Monsoon"
    else:
        return "Autumn"

df["Season"] = df["Month"].apply(get_season)

display(df.head())
```

```python
print("First 5 rows:")
display(df.head())

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
total_revenue = df["Revenue"].sum()
total_cost = df["Cost"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order_ID"].nunique()
total_units = df["Quantity"].sum()

profit_margin = (
    total_profit / total_revenue * 100
    if total_revenue else 0
)

print("========== BUSINESS KPIs ==========")
print(f"Total Revenue: PKR {total_revenue:,.2f}")
print(f"Total Cost: PKR {total_cost:,.2f}")
print(f"Total Profit: PKR {total_profit:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Units Sold: {total_units:,}")
print(f"Profit Margin: {profit_margin:.2f}%")
```

```python
monthly_sales = (
    df.groupby(["Month", "Month_Name"], as_index=False)
      .agg(
          Revenue=("Revenue", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order_ID", "nunique"),
          Units_Sold=("Quantity", "sum")
      )
      .sort_values("Month")
)

display(monthly_sales.round(2))

highest_month = monthly_sales.loc[
    monthly_sales["Revenue"].idxmax()
]

lowest_month = monthly_sales.loc[
    monthly_sales["Revenue"].idxmin()
]

print(
    f"Highest-sales month: {highest_month['Month_Name']} "
    f"(PKR {highest_month['Revenue']:,.2f})"
)

print(
    f"Lowest-sales month: {lowest_month['Month_Name']} "
    f"(PKR {lowest_month['Revenue']:,.2f})"
)
```

```python
season_order = [
    "Winter", "Spring", "Summer", "Monsoon", "Autumn"
]

seasonal_sales = (
    df.groupby("Season", as_index=False)
      .agg(
          Revenue=("Revenue", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order_ID", "nunique"),
          Units_Sold=("Quantity", "sum")
      )
)

seasonal_sales["Season"] = pd.Categorical(
    seasonal_sales["Season"],
    categories=season_order,
    ordered=True
)

seasonal_sales = seasonal_sales.sort_values("Season")

display(seasonal_sales.round(2))

highest_season = seasonal_sales.loc[
    seasonal_sales["Revenue"].idxmax()
]

lowest_season = seasonal_sales.loc[
    seasonal_sales["Revenue"].idxmin()
]

print("Highest-sales season:", highest_season["Season"])
print("Lowest-sales season:", lowest_season["Season"])
```

```python
product_sales = (
    df.groupby("Product", as_index=False)
      .agg(
          Revenue=("Revenue", "sum"),
          Profit=("Profit", "sum"),
          Units_Sold=("Quantity", "sum")
      )
      .sort_values("Revenue", ascending=False)
)

display(product_sales.round(2))

print("Highest-revenue product:", product_sales.iloc[0]["Product"])
print("Lowest-revenue product:", product_sales.iloc[-1]["Product"])
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_sales["Month_Name"],
    monthly_sales["Revenue"],
    marker="o"
)

plt.title("Monthly Grocery Store Revenue - 2025")
plt.xlabel("Month")
plt.ylabel("Revenue (PKR)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    seasonal_sales["Season"].astype(str),
    seasonal_sales["Revenue"]
)

plt.title("Grocery Revenue by Season")
plt.xlabel("Season")
plt.ylabel("Revenue (PKR)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    product_sales["Product"],
    product_sales["Revenue"]
)

plt.title("Revenue by Grocery Product")
plt.xlabel("Product")
plt.ylabel("Revenue (PKR)")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    product_sales["Product"],
    product_sales["Profit"]
)

plt.title("Profit by Grocery Product")
plt.xlabel("Product")
plt.ylabel("Profit (PKR)")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```
